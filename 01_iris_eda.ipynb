# **1. Построить гистограмму распределения длины лепестка.**

```python
import matplotlib.pyplot as plt

# Подсчёт количества объектов каждого класса
class_counts = y.value_counts().sort_index()

# Замена числовых кодов названиями классов
class_counts.index = iris.target_names

# Построение диаграммы
plt.figure(figsize=(8, 5))

plt.bar(
    class_counts.index,
    class_counts.values
)

plt.title("Распределение классов Iris")
plt.xlabel("Вид ириса")
plt.ylabel("Количество объектов")

plt.show()
```

# **2. Построить столбчатую диаграмму количества объектов каждого класса.**

```python
plt.figure(figsize=(8, 5))

plt.hist(
    X["petal length (cm)"],
    bins=15,
    edgecolor="black"
)

plt.title("Распределение длины лепестков Iris")
plt.xlabel("Длина лепестка (см)")
plt.ylabel("Количество наблюдений")

plt.show()
```
